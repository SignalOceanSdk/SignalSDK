## Call the Vessel Consumptions API
The Vessel Consumptions API retrieves information about vessel consumptions and advertised consumptions for vessels.

````python
from signal_ocean import Connection
from signal_ocean.vessel_consumptions import VesselConsumptionsAPI

import pandas as pd

connection = Connection(signal_ocean_api_key)
api = VesselConsumptionsAPI(connection)
````

## Examples

### Get Consumptions for a vessel

````python
vessel_consumptions = api.get_consumptions(imo=9412036)
print("Return type is:", type(vessel_consumptions), "\n")
vessel_consumptions
````

#### Note that this function returns a custom object. To get a dictionary object, as in the public API response, we can call class method _to_dict()_.

````python
vessel_consumptions.to_dict()
````

### View Laden Consumptions

````python
if vessel_consumptions.laden_consumptions:
    laden_df = pd.DataFrame([c.__dict__ for c in vessel_consumptions.laden_consumptions])
    laden_df
````

### View Ballast Consumptions

````python
if vessel_consumptions.ballast_consumptions:
    ballast_df = pd.DataFrame([c.__dict__ for c in vessel_consumptions.ballast_consumptions])
    ballast_df
````

### Get Advertised Consumptions for a vessel

````python
advertised_consumptions = api.get_advertised_consumptions(imo=9412036)
print(f"Return type is: {type(advertised_consumptions)}")
advertised_consumptions
````

#### Note that this function returns a custom object. To get a dictionary object, as in the public API response, we can call method _to_dict()_.

## Get Advertised Consumptions (Paginated)

This endpoint retrieves all advertised consumptions for all vessels in a paginated format.

### Get first page of advertised consumptions

````python
page = api.get_advertised_consumptions_paginated(page_size=10)
print(f"Return type is: {type(page)}")
print(f"Number of results: {len(page.data)}")
print(f"Next page token: {page.next_page_token}")
````

### Convert first result to dictionary

````python
if page.data:
    page.data[0].to_dict()
````

### Get next page using the token

````python
if page.next_page_token:
    next_page = api.get_advertised_consumptions_paginated(
        token=page.next_page_token,
        page_size=10
    )
    print(f"Number of results in next page: {len(next_page.data)}")
````

### Convert paginated results to DataFrame

We can convert the paginated response to a DataFrame by extracting the data and using pd.json_normalize.

````python
page_dict = page.to_dict()
df = pd.json_normalize(page_dict['Data'], sep='')
df.head()
````
